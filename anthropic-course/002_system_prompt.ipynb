# Create an API client

```python
import os

from anthropic import Anthropic

client = Anthropic(api_key=os.getenv("ANTHROPIC_API_KEY"))
model = "claude-sonnet-4-6"
```

# Helper functions

```python
def add_user_message(messages, user_message):
    messages.append({
        "role": "user",
        "content": user_message
    })

def add_assistant_message(messages, assistant_message):
    messages.append({
        "role": "assistant",
        "content": assistant_message
    })

def chat(messages, system):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }
    if system:
        params["system"] = system
    try:
        message = client.messages.create(
            **params
        )
        return message.content[0].text
    


    except Exception as e:
        print("An error occurred:")
        print(type(e).__name__)
        return e
```

# Make a request

```python
messages = []
system = """
You're a expert in python who writes very concise code.
"""
add_user_message(messages, "Write a function to find duplicate strings in a list")
answer = chat(messages, system)
```
