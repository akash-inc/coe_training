## Create an API client

```python
import os

from anthropic import Anthropic

client = Anthropic(api_key=os.getenv("ANTHROPIC_API_KEY"))
model = "claude-sonnet-4-6"
```

## Helper functions

```python
def add_user_message(messages, user_message):
    messages.append({
        "role": "user",
        "content": user_message
    })

def add_assistant_message(messages, assistant_message):
    messages.append({
        "role": "assistant",
        "content": assistant_message
    })

EXIT_PHRASES = frozenset({"exit", "quit", "q"})


def get_assistant_reply(messages):
    try:
        message = client.messages.create(
            model=model,
            max_tokens=1000,
            messages=messages
        )
        return message.content[0].text
    except Exception as e:
        print("An error occurred:")
        print(type(e).__name__)
        return e


def prompt_user_turn(messages):
    while True:
        text = input("You: ").strip()
        lowered = text.lower()
        if lowered in EXIT_PHRASES:
            return False
        if text:
            add_user_message(messages, text)
            print(text)
            return True
        print("Say something—or type exit to quit.")


def run_assistant_turn(messages):
    answer = get_assistant_reply(messages)
    if isinstance(answer, Exception):
        print(answer)
        add_assistant_message(messages, "Request failed.")
        return
    add_assistant_message(messages, answer)
    print(answer)

```

## Make a request

```python
messages = []
while prompt_user_turn(messages):
    run_assistant_turn(messages)
```
