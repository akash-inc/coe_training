```python
from dotenv import load_dotenv
from anthropic import Anthropic
import os

load_dotenv()

client = Anthropic(api_key=os.getenv("ANTHROPIC_API_KEY"))
model = "claude-sonnet-4-6"
```

```python
answer = client.messages.create(
    model=model,
    max_tokens=500,
    messages=[
        {"role": "user", "content": "How do we solve the problem of duplicate strings in a list?"}
    ],
    system="You're a expert in python who writes very concise code.",
    temperature=0.2,
    stream=True
)

for event in answer:
    print(event)
```

```python
with client.messages.stream(
    model=model,
    max_tokens=500,
    messages=[
        {"role": "user", "content": "How do we solve the problem of duplicate strings in a list?"}
    ],
    system="You're a expert in python who writes very concise code.",
) as stream:
    for text in stream.text_stream:
        pass
        # print(text, end="")

answer = stream.get_final_text()
print(answer)

```
